# Modules

Following the principle of modularity, Python allows to organize code into separate files called modules. A module is simply a file with a `.py` extension that contains Python code. Modules can define functions, classes, and variables that can be reused in other Python programs. Several modules can be grouped together into a package, which is a directory containing a special `__init__.py` file. This allows for better organization and namespace management. These packages can be installed from the Python Package Index (PyPI) using package managers like `pip`. Python has over 200,000 registered packages, and so a big part of becoming proficient has to do with importing third-party libraries.

(importing)=
## Importing Packages

To import a package, you can use the `import` statement. For example, to import the entire [numpy](https://numpy.org/) package and assign a short-hand alias for it to keep your code more concise, you can use:
```
import numpy as np
```

````python
import numpy as np
````

Sub-modules can be accessed using a `.` such that

````python
print(np.array)
````

returns a function handle for the `array` class.
