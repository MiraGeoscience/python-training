# Custom User Interface (ui.json)


This section provides training on how to create custom user interface within Geoscience ANALYST using the [ui.json
standard](https://geoh5py.readthedocs.io/en/stable/content/uijson_format/index.html). In this section we are going to demonstrate how to:

- Create a basic user-interface programmatically
- Read and write a ui.json from a file
- Demonstrate how to set up ANALYST to run a custom program

## Creating a ui.json

The `ui.json` format is based on the [json](https://www.json.org/json-en.html) standard - a format widely used in programming. Python comes by default with a [json](https://docs.python.org/3/library/json.html) module to read and write to files. We then need a way to create the structure that ANALYST understands. We will make use of the [geoh5py.ui_json](https://mirageoscience-geoh5py.readthedocs-hosted.com/en/stable/content/uijson_format/index.html) module to get started.

First, we need the base components must contain a few base parameters, .

The `UIJson` class has been created to easily read and write `ui.json` files. It leverages [pydantic](https://pydantic.dev/docs/validation/dev/get-started/) to validate the inputs. We can generate a simple UI with the minimum requirements, such as a `title` and a `run_command`:

```python
core_parameters = {
    "title": "Hello World",
    "geoh5": "",
    "run_command": "",
    "conda_environment": "",
}
```

Note that the `run_command` is currently set to `""`, and therefore nothing will be executed by ANALYST. We will get to that part later.
This user interface does nothing at the moment. We need to add additional forms to capture inputs from the user.

### Forms

There are many types of forms that can be added. See the [Forms documentation](https://mirageoscience-geoh5py.readthedocs-hosted.com/en/stable/content/uijson_format/forms.html) for an exhaustive list.

For example, we can add a simple `BoolForm` that defines a checkbox option. Programmatically, we can create this UI with a custom UIJson class

```python
class MyUIJson(UIJson):
    bool_form: forms.BoolForm


my_ui = MyUIJson(
    bool_form=forms.BoolForm(label="Check me", value=True, main=True), **core_parameters
)
```
