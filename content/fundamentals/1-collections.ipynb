# Collections

In the previous section, we have introduced numerics of type `int` and `float`, as well as text variable. Often we would like to group many of numerics or strings together to perform larger operations. This section introduces classes of ``collections`` to group values together.

## List

Lists are generic containers for other objects, which can be of any type. They are created with square brackets `[]`.

```python
a_list = [1.0, 2, "abc"]
```

Here are a few important methods that can used on a `list`.

- Indexing (accessing) elements of the list (count starts at 0)

```python
a_list[0]
```

- Adding (appending) to the list.

```python
a_list += ["new"]
a_list
```

- Removing an element from the list using the `pop` method with index

```python
a_list.pop(0)
a_list
```

- Finding the `index` of an element within the list. The same method applies to [string](#strings) objects.

```python
a_list.index("abc")
```

## Tuple

Tuples are similar to lists but are immutable, meaning that they cannot be modified after creation. They are created with parentheses `()`.

```python
a_tuple = (1.0, 2, "abc")

try:
    a_tuple[0] = 2.0  # This will raise an error
except TypeError as e:
    print(f"Error: {e}")
```

## Dictionary

Dictionaries (`dict`) are also containers for objects but give an additional level of structure over lists as every
value of the dictionary is indexed by a `key`. A dictionary is created with a list of `key: value` pairs within
braces `{}`.

```python
my_dict = {"integer": 2}
```

Here are a few important methods that can used on a `dict`.

- Accessing values using the key

```python
my_dict["integer"]
```

- Adding entries

```python
my_dict["new"] = 123
my_dict
```

- Removing entries

```python
del my_dict["new"]
my_dict
```

- Looping over keys

```python
list(my_dict)
```

- Looping over values

```python
list(my_dict.values())
```

- Looping over both keys and values

```python
list(my_dict.items())
```

## Numpy Arrays

[Numpy](https://numpy.org/) is a Python library that provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays. It is widely used in scientific computing and data analysis. This section introduces the basic concepts of `numpy` arrays and a handful of operations that will be used in later sections.

We can create a 2D array with the `np.array()` method, and then perform operations on it. For example, we can create a 2D array and compute the sum along the first axis (rows).

```python
import numpy as np

array = np.array([[1, 2, 3], [4, 5, 6]])  # Create a 2D array
array.sum(axis=0)
```

Many times we will need to perform operations on two arrays. For example, we can compute the dot product of two arrays.

```python
np.dot(array, array.T)  # Compute the dot product
```

We use the transpose (`.T`) method allows to align the dimensions of the arrays for the dot product.

$[1 \times 3] \cdot [M \times 3].T \rightarrow [1 \times M] $.

Array broadcasting allows to operate with arrays of different dimensions using the `None` index:

$[N \times 3] \;/\;([N][None, :]) \rightarrow [N \times 3] $

```python
array / np.array([1, 2, 3])[None, :]  # Broadcast a 1D array along a second dimension
```
